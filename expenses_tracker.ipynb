```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("expenses.csv")

# Convert date column
df["date"] = pd.to_datetime(df["date"])

# Create month column
df["month"] = df["date"].dt.month_name()

# Separate expenses
expenses = df.loc[df["type"] == "Expense"]

# Total income and expenses
tot_inc = df.loc[df["type"] == "Income", "amount"].sum()
tot_exp = expenses["amount"].sum()

# Savings
savings = tot_inc - tot_exp

# Savings percentage
savings_percentage = (savings / tot_inc) * 100

# Expense percentage
percentage_spent = (tot_exp / tot_inc) * 100

# Total transactions
total_transactions = df.shape[0]
total_expense_transactions = expenses.shape[0]
total_income_transactions = df.loc[
    df["type"] == "Income"
].shape[0]

# Spending by category
expense_by_category = (
    expenses.groupby("category")["amount"].sum()
)

# Highest spending category
highest_category = expense_by_category.idxmax()
highest_amount = expense_by_category.max()

# Average and median expense
average_expense = np.mean(expenses["amount"])
median_expense = np.median(expenses["amount"])

# Most frequent category
category_counts = expenses["category"].value_counts()
most_frequent_category = category_counts.idxmax()

# Expense percentage by category
expense_percentage = (
    expense_by_category / tot_exp
) * 100

# Spending by payment method
expense_by_payment = (
    expenses.groupby("payment_method")["amount"].sum()
)

# Monthly income and expenses
monthly_income = (
    df.loc[df["type"] == "Income"]
      .groupby("month")["amount"]
      .sum()
)

monthly_expenses = (
    expenses.groupby("month")["amount"].sum()
)

# Largest and smallest transactions
largest_expense = expenses.loc[
    expenses["amount"].idxmax()
]

smallest_expense = expenses.loc[
    expenses["amount"].idxmin()
]

# Difference between largest and smallest expense
expense_difference = (
    expenses["amount"].max()
    - expenses["amount"].min()
)

# Number of transactions by category
category_transaction_count = (
    expenses["category"].value_counts()
)

# Average expense by category
average_by_category = (
    expenses.groupby("category")["amount"].mean()
)

# Final monthly DataFrame
monthly_data = pd.DataFrame({
    "Income": monthly_income,
    "Expense": monthly_expenses
})

# Correlation
correlation = monthly_data["Income"].corr(
    monthly_data["Expense"]
)

# Final report
print("PERSONAL FINANCE SUMMARY")
print("=" * 30)

print("Total income:", tot_inc)
print("Total expenses:", tot_exp)
print("Savings:", savings)
print("Savings percentage:", savings_percentage)
print("Percentage of income spent:", percentage_spent)

print("\nTRANSACTION DETAILS")
print("-" * 30)
print("Total transactions:", total_transactions)
print("Income transactions:", total_income_transactions)
print("Expense transactions:", total_expense_transactions)

print("\nCATEGORY DETAILS")
print("-" * 30)
print("Highest spending category:", highest_category)
print("Highest category amount:", highest_amount)
print("Most frequent category:", most_frequent_category)

print("\nEXPENSE STATISTICS")
print("-" * 30)
print("Average expense:", average_expense)
print("Median expense:", median_expense)
print("Largest-smallest difference:", expense_difference)

print("\nLARGEST EXPENSE")
print("-" * 30)
print(largest_expense)

print("\nSMALLEST EXPENSE")
print("-" * 30)
print(smallest_expense)

print("\nEXPENSE BY CATEGORY")
print("-" * 30)
print(expense_by_category)

print("\nEXPENSE PERCENTAGE BY CATEGORY")
print("-" * 30)
print(expense_percentage)

print("\nEXPENSE BY PAYMENT METHOD")
print("-" * 30)
print(expense_by_payment)

print("\nMONTHLY DATA")
print("-" * 30)
print(monthly_data)

print("\nCORRELATION")
print("-" * 30)
print(correlation)
```
